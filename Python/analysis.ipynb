```python
# Convert date columns to datetime

df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print(df[["Order Date", "Ship Date"]].head())
```

```python
# Analyze the relationship between discount and profit

discount_profit = (
    df.groupby("Discount")["Profit"]
      .mean()
      .sort_index()
)

print(discount_profit)
```

```python
# Discount vs Average Profit

discount_profit.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Average Profit by Discount Level")
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.xticks(rotation=45)
plt.axhline(y=0, linestyle="--")
plt.tight_layout()
plt.show()
```

```python
# Number of records at each discount level

discount_count = (
    df.groupby("Discount")
      .size()
      .sort_index()
)

print(discount_count)
```

```python
# Create discount groups

def discount_group(discount):
    if discount == 0:
        return "0%"
    elif discount <= 0.20:
        return "1-20%"
    elif discount <= 0.40:
        return "21-40%"
    elif discount <= 0.60:
        return "41-60%"
    else:
        return "61-80%"


df["Discount Group"] = df["Discount"].apply(discount_group)

print(df["Discount Group"].value_counts())
```

```python
# Average profit by discount group

discount_group_profit = (
    df.groupby("Discount Group")["Profit"]
      .mean()
)

print(discount_group_profit)
```

```python
# Average profit by discount group

discount_group_profit.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Average Profit by Discount Group")
plt.xlabel("Discount Group")
plt.ylabel("Average Profit")
plt.axhline(y=0, linestyle="--")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Identify loss-making records

loss_data = df[df["Profit"] < 0]

print("Loss-making records:", len(loss_data))
print("Total Loss:", round(loss_data["Profit"].sum(), 2))
```

```python
# Total loss by category

loss_by_category = (
    loss_data.groupby("Category")["Profit"]
             .sum()
             .sort_values()
)

print(loss_by_category)
```

```python
# Visualize total loss by category

loss_by_category.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Total Loss by Category")
plt.xlabel("Category")
plt.ylabel("Total Loss")
plt.axhline(y=0, linestyle="--")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Total loss by sub-category

loss_by_subcategory = (
    loss_data.groupby("Sub-Category")["Profit"]
             .sum()
             .sort_values()
)

print(loss_by_subcategory)
```

```python
# Top 10 loss-making sub-categories

top_loss_subcategories = loss_by_subcategory.head(10)

top_loss_subcategories.plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 Loss-Making Sub-Categories")
plt.xlabel("Total Loss")
plt.ylabel("Sub-Category")
plt.axvline(x=0, linestyle="--")
plt.tight_layout()
plt.show()
```

```python
# Top 10 loss-making products

loss_by_product = (
    loss_data.groupby("Product Name")["Profit"]
             .sum()
             .sort_values()
)

top_loss_products = loss_by_product.head(10)

print(top_loss_products)
```

```python
# Top 10 loss-making products

top_loss_products.sort_values().plot(
    kind="barh",
    figsize=(11, 7)
)

plt.title("Top 10 Loss-Making Products")
plt.xlabel("Total Loss")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
# Analyze discount for top loss-making products

top_loss_product_discount = (
    loss_data[loss_data["Product Name"].isin(top_loss_products.index)]
    .groupby("Product Name")["Discount"]
    .mean()
    .sort_values(ascending=False)
)

print(top_loss_product_discount)
```

```python
# Check number of records for top loss-making products

top_loss_product_count = (
    loss_data[loss_data["Product Name"].isin(top_loss_products.index)]
    .groupby("Product Name")
    .size()
    .sort_values(ascending=False)
)

print(top_loss_product_count)
```

```python
# Find products with at least 10 loss-making records

product_loss_summary = (
    loss_data.groupby("Product Name")
    .agg(
        Loss=("Profit", "sum"),
        Average_Profit=("Profit", "mean"),
        Average_Discount=("Discount", "mean"),
        Loss_Records=("Profit", "count")
    )
)

reliable_loss_products = (
    product_loss_summary[
        product_loss_summary["Loss_Records"] >= 10
    ]
    .sort_values("Loss")
)

print(reliable_loss_products.head(10))
```

```python
# Analyze loss rate for products

product_summary = (
    df.groupby("Product Name")
    .agg(
        Total_Records=("Profit", "count"),
        Total_Profit=("Profit", "sum"),
        Average_Profit=("Profit", "mean"),
        Average_Discount=("Discount", "mean")
    )
)

product_summary["Loss_Records"] = (
    df[df["Profit"] < 0]
    .groupby("Product Name")["Profit"]
    .count()
)

product_summary["Loss_Records"] = (
    product_summary["Loss_Records"].fillna(0)
)

product_summary["Loss_Rate"] = (
    product_summary["Loss_Records"]
    / product_summary["Total_Records"]
    * 100
)

repeated_loss_products = (
    product_summary[
        product_summary["Loss_Records"] >= 10
    ]
    .sort_values("Loss_Rate", ascending=False)
)

print(
    repeated_loss_products[
        [
            "Total_Records",
            "Loss_Records",
            "Loss_Rate",
            "Total_Profit",
            "Average_Discount"
        ]
    ].head(10)
)
```

```python
# Analyze sales and profit for repeated loss-making products

loss_product_analysis = (
    df[df["Product Name"].isin(repeated_loss_products.head(10).index)]
    .groupby("Product Name")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Average_Discount=("Discount", "mean"),
        Total_Records=("Profit", "count")
    )
    .sort_values("Total_Profit")
)

print(loss_product_analysis)
```

```python
# Calculate profit margin for repeated loss-making products

loss_product_analysis["Profit_Margin"] = (
    loss_product_analysis["Total_Profit"]
    / loss_product_analysis["Total_Sales"]
    * 100
)

print(
    loss_product_analysis[
        [
            "Total_Sales",
            "Total_Profit",
            "Profit_Margin",
            "Average_Discount"
        ]
    ]
)
```

```python
# Profit margin of repeatedly loss-making products

profit_margin = (
    loss_product_analysis["Profit_Margin"]
    .sort_values()
)

profit_margin.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Profit Margin of Repeatedly Loss-Making Products")
plt.xlabel("Product")
plt.ylabel("Profit Margin (%)")
plt.axhline(y=0, linestyle="--")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
