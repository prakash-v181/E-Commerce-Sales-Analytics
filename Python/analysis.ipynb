```python
print("Total Sales :", round(df["Sales"].sum(), 2))
print("Total Profit:", round(df["Profit"].sum(), 2))
print("Total Orders:", len(df))
print("Average Sales:", round(df["Sales"].mean(), 2))
```

```python
import os
import matplotlib.pyplot as plt

# Create the charts folder if it doesn't exist
os.makedirs("charts", exist_ok=True)

category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")

plt.title("Sales by Category", fontsize=16, fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.savefig("charts/sales_by_category.png", dpi=300)

plt.show()
```

```python
import os
import matplotlib.pyplot as plt

# Create the charts folder if it doesn't exist
os.makedirs("charts", exist_ok=True)

# Calculate sales by category
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

# Create the chart
ax = category_sales.plot(kind="bar", figsize=(8,5))

# Add title and labels
ax.set_title("Sales by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Total Sales")

plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Adjust layout
plt.tight_layout()

# Save chart
plt.savefig("charts/sales_by_category.png", dpi=300)

# Display chart
plt.show()
```

```python
import matplotlib.pyplot as plt

region_profit = (
    df.groupby("Region")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

fig, ax = plt.subplots(figsize=(8,5))

region_profit.plot(kind="bar", ax=ax)

ax.set_title("Profit by Region", fontsize=16, fontweight="bold")
ax.set_xlabel("Region")
ax.set_ylabel("Total Profit ($)")

plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

fig.savefig("charts/profit_by_region.png", dpi=300)

plt.show()
```

```python
import matplotlib.pyplot as plt

top_customers = (
    df.groupby("Customer Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

fig, ax = plt.subplots(figsize=(10,6))

top_customers.plot(kind="barh", ax=ax)

ax.set_title("Top 10 Customers by Sales", fontsize=16, fontweight="bold")
ax.set_xlabel("Total Sales ($)")
ax.set_ylabel("Customer")

plt.tight_layout()

fig.savefig("charts/top10_customers.png", dpi=300)

plt.show()
```

```python
import matplotlib.pyplot as plt

top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

fig, ax = plt.subplots(figsize=(12,6))

top_products.plot(kind="barh", ax=ax)

ax.set_title("Top 10 Products by Sales", fontsize=16, fontweight="bold")
ax.set_xlabel("Total Sales ($)")
ax.set_ylabel("Product")

plt.tight_layout()

fig.savefig("charts/top10_products.png", dpi=300)

plt.show()
```
